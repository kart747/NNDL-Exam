# IMDB sentiment with a SimpleRNN

Embedding plus SimpleRNN on the Keras IMDB dataset (10,000-word vocabulary, sequence length 100).


```python
import tensorflow as tf
# Allocate GPU memory only as needed. By default TensorFlow grabs all of it, which makes a second
# notebook (or kernel) fail with an out-of-memory error on the 4 GB laptop GPU.
for gpu in tf.config.list_physical_devices('GPU'):
    tf.config.experimental.set_memory_growth(gpu, True)
```

```python
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
(X_train, y_train), (X_test, y_test) = keras.datasets.imdb.load_data(num_words=10000)
```

```python
X_train = pad_sequences(X_train,maxlen=100)
X_test = pad_sequences(X_test,maxlen=100)
```

```python
model = keras.Sequential([
    layers.Embedding(10000,32),
    layers.SimpleRNN(32),
    layers.Dense(1, activation='sigmoid')
])
```

```python
model.compile(
    optimizer='rmsprop',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
model.fit(
    X_train, y_train,
    epochs=10,
    validation_split=0.2
)
```

```python
print('Test accuracy:', model.evaluate(X_test, y_test)[1])
```
