```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from tensorflow import keras
from tensorflow.keras.layers import Dense, Dropout

X, y = make_moons(100, noise=0.25, random_state=2)

m1 = keras.Sequential([Dense(128, activation='relu'), Dense(128, activation='relu'), Dense(1, activation='sigmoid')])
m2 = keras.Sequential([Dense(128, activation='relu'), Dropout(0.2), Dense(128, activation='relu'), Dropout(0.2), Dense(1, activation='sigmoid')])
m3 = keras.Sequential([Dense(128, activation='relu'), Dropout(0.5), Dense(128, activation='relu'), Dropout(0.5), Dense(1, activation='sigmoid')])

m1.compile('adam', 'binary_crossentropy', metrics=['accuracy'])
m2.compile('adam', 'binary_crossentropy', metrics=['accuracy'])
m3.compile('adam', 'binary_crossentropy', metrics=['accuracy'])

h1 = m1.fit(X, y, epochs=500, validation_split=0.2, verbose=0)
h2 = m2.fit(X, y, epochs=500, validation_split=0.2, verbose=0)
h3 = m3.fit(X, y, epochs=500, validation_split=0.2, verbose=0)

print('Val acc:', h1.history['val_accuracy'][-1], h2.history['val_accuracy'][-1], h3.history['val_accuracy'][-1])
plt.plot(h1.history['val_loss']); plt.plot(h2.history['val_loss']); plt.plot(h3.history['val_loss'])
plt.legend(['No dropout', 'Dropout 0.2', 'Dropout 0.5']); plt.show()
```

```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from tensorflow import keras
from tensorflow.keras.layers import Dense, Dropout
from tensorflow.keras.regularizers import l1, l2

X, y = make_moons(100, noise=0.25, random_state=2)

m1 = keras.Sequential([Dense(128, activation='relu'), Dense(128, activation='relu'), Dense(1, activation='sigmoid')])
m2 = keras.Sequential([Dense(128, activation='relu', kernel_regularizer=l1(0.001)), Dense(128, activation='relu', kernel_regularizer=l1(0.001)), Dense(1, activation='sigmoid')])
m3 = keras.Sequential([Dense(128, activation='relu', kernel_regularizer=l2(0.001)), Dense(128, activation='relu', kernel_regularizer=l2(0.001)), Dense(1, activation='sigmoid')])
m4 = keras.Sequential([Dense(128, activation='relu'), Dropout(0.5), Dense(128, activation='relu'), Dropout(0.5), Dense(1, activation='sigmoid')])

m1.compile('adam', 'binary_crossentropy', metrics=['accuracy'])
m2.compile('adam', 'binary_crossentropy', metrics=['accuracy'])
m3.compile('adam', 'binary_crossentropy', metrics=['accuracy'])
m4.compile('adam', 'binary_crossentropy', metrics=['accuracy'])

h1 = m1.fit(X, y, epochs=500, validation_split=0.2, verbose=0)
h2 = m2.fit(X, y, epochs=500, validation_split=0.2, verbose=0)
h3 = m3.fit(X, y, epochs=500, validation_split=0.2, verbose=0)
h4 = m4.fit(X, y, epochs=500, validation_split=0.2, verbose=0)

print('Val acc:', h1.history['val_accuracy'][-1], h2.history['val_accuracy'][-1],
      h3.history['val_accuracy'][-1], h4.history['val_accuracy'][-1])
plt.plot(h1.history['val_loss']); plt.plot(h2.history['val_loss'])
plt.plot(h3.history['val_loss']); plt.plot(h4.history['val_loss'])
plt.legend(['None', 'L1', 'L2', 'Dropout']); plt.show()
```
