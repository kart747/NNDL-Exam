# MNIST with a two-hidden-layer MLP

Flatten, Dense(128), Dense(64), softmax on MNIST. Trains for 10 epochs and prints actual vs. predicted digits.


```python
import tensorflow as tf
# Allocate GPU memory only as needed. By default TensorFlow grabs all of it, which makes a second
# notebook (or kernel) fail with an out-of-memory error on the 4 GB laptop GPU.
for gpu in tf.config.list_physical_devices('GPU'):
    tf.config.experimental.set_memory_growth(gpu, True)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from tensorflow import keras

(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()
```

```python
X_train, X_test = X_train / 255.0, X_test / 255.0
X_train[0].shape
```

```python
model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(64, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])
```

```python
model.compile(optimizer='adam',loss='sparse_categorical_crossentropy',metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, epochs=10,
          batch_size=128, validation_split=0.1)
```

```python
print(model.evaluate(X_test, y_test))
```

```python
y_pred = np.argmax(model.predict(X_test),axis=1)
```

```python
for i in range(10):
    print('Actual:', y_test[i], 'Predicted:', y_pred[i])
```

```python
plt.imshow(X_test[0], cmap='gray')
plt.title(f'Actual: {y_test[0]} Predicted: {y_pred[0]}')
plt.axis('off')
plt.show()

# Confusion matrix
ConfusionMatrixDisplay(
    confusion_matrix(y_test, y_pred)
).plot()
plt.show()
```
