# MNIST MLP with a confusion matrix

A smaller one-hidden-layer MLP on MNIST, evaluated with a confusion matrix.


```python
import tensorflow as tf
# Allocate GPU memory only as needed. By default TensorFlow grabs all of it, which makes a second
# notebook (or kernel) fail with an out-of-memory error on the 4 GB laptop GPU.
for gpu in tf.config.list_physical_devices('GPU'):
    tf.config.experimental.set_memory_growth(gpu, True)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import ConfusionMatrixDisplay
from tensorflow import keras

(X_train, y_train), (X_test, y_test) = keras.datasets.mnist.load_data()
X_train, X_test = X_train / 255.0, X_test / 255.0

model = keras.Sequential([
    keras.layers.Flatten(input_shape=(28, 28)),
    keras.layers.Dense(128, activation='relu'),
    keras.layers.Dense(10, activation='softmax')
])
model.compile(optimizer='adam', loss='sparse_categorical_crossentropy', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=5)
print(model.evaluate(X_test, y_test))

y_pred = np.argmax(model.predict(X_test), axis=1)
plt.imshow(X_test[0], cmap='gray'); plt.title(f'Actual: {y_test[0]}  Predicted: {y_pred[0]}'); plt.show()
ConfusionMatrixDisplay.from_predictions(y_test, y_pred); plt.show()
```
